```python
E1=np.random.exponential(scale=1,size=100000)
E2=np.random.exponential(scale=2.5,size=100000)
E3=np.random.exponential(scale=8.5,size=100000)
```

```python
import matplotlib.pyplot as plt
```

```python
sb.histplot(E1,alpha=1,label=r'$\lambda$=1')
sb.histplot(E2,alpha=0.6,label=r'$\lambda$=0.5')
sb.histplot(E3,alpha=0.3,label=r'$\lambda$=2.5')
plt.legend()
```

```python
print(np.mean(E1))
print(np.mean(E2))
print(np.mean(E3))
print(np.mean(E4))
print(np.mean(E5))
```

```python
print(np.var(E1))
print(np.var(E2))
print(np.var(E3))
print(np.var(E4))
print(np.var(E5))
```
