```python
from scipy.integrate import quad
```

```python
import numpy as np
```

$$\int_0^1 (ax^2+b)\mathrm{d}x$$


```python
def I(x,a,b):
  return a*x**2+b
```

```python
a=2
b=1
```

```python
quad(I,0,1,args=(a,b))
```

$$\int_1^\infty\frac{e^{-xt}}{t^n} \mathrm{d}t$$

```python
def f(t,n,x):
  return np.exp(-x*t) / t**n
```

```python
#n=4
#x=np.pi
quad(f,1,np.inf,args=(n,x))[0]
```

```python
def integralf(n,x):
  return quad(f,1,np.inf,args=(n,x))[0]
```

```python
vecf=np.vectorize(integralf)
```

$$\int_1^\infty\int_0^{\infty}\frac{e^{-xt}}{t^n} \mathrm{d}t\mathrm{d}x$$

```python
vecf(3,np.arange(1,4,0.5))
```

```python
quad(lambda x: integralf(3,x),0,np.inf)
```

```python
from scipy.integrate import  dblquad
```

```python
def I(n):
  return np.round(dblquad(lambda t,x: np.exp(-x*t)/t**n,0,np.inf,lambda x: 1,lambda x: np.inf)[0],2)
```

```python
I(3)
```

```python
I(8)
```

```python
I(100)
```
