```python
import numpy as np
import pandas as pd
import seaborn as sb
```

```python
enigh=pd.read_csv('/content/concentradohogar.csv')
```

```python
data=pd.read_csv('/content/AGEEML_20265211513372.csv')
```

```python
data
```

```python
enigh=enigh.rename(columns={'ubica_geo':'CVEGEO'})
```

```python
enigh=enigh.merge(data,on='CVEGEO')
```

```python
enigh
```

```python
enigh.keys().values
```

```python
enigh2=enigh.filter(['CVEGEO','sexo_jefe','edad_jefe','educa_jefe','ing_cor','ingtrab','jubilacion','becas','remesas','bene_gob','NOM_ENT','NOM_MUN'])
```

```python
enigh2
```

```python
enigh2=enigh2.rename(columns={'NOM_ENT':'estado','NOM_MUN':'municipio'})
```

```python
enigh2
```

```python
Tabla1=pd.pivot_table(enigh2,index=['estado'],columns=['sexo_jefe'],values=['ing_cor'],aggfunc='mean')
#
```

```python
Tabla1=Tabla1.rename(columns={1:'hombres',2:'mujeres'})
```

```python
Tabla1.round(2).to_excel('T1.xlsx')
```
